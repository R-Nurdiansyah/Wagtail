```python
file_map = pl.read_csv('file_map_EMP1.tsv', separator = '\t', has_header=False)
file_map[:4]
```

```python
file = 'file_map_EMP1.tsv'
```

```python
filename = [line.strip() for line in file]
print(filename)
```

```python
#make dictionary from file_map column_1 and column_2
file_map_dict = dict(zip(file_map['column_1'], file_map['column_2']))
file_map_dict
```

```python
input = "ERR4994173"
output = "./"
```

```python
import os
```

```python
if input in file_map_dict:
    forward_file = file_map_dict[input]
    df_manifest = pl.DataFrame({
        "sample-id": [input],
        "absolute-filepath": [os.path.abspath(forward_file)],
        "direction": ["forward"]
    })
    manifest_file = os.path.join(output, input + "_manifest.csv")
    df_manifest.write_csv(manifest_file, has_header=True)
    print("Created manifest file for sample:", input)
```
